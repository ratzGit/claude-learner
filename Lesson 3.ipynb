```python
# Load headers

from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
client = Anthropic()
model ="claude-sonnet-4-0"
```

```python
# Helper functions

def add_user_messages(messages, text):
    user_message = {"role":"user", "content": text}
    messages.append(user_message)


def add_assistant_messages(messages, text):
    assistant_message = {"role":"assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    
    params = {
        "model": model,
        "max_tokens":1000,
        "messages": messages,
        "temperature":temperature
    }

    if system:
        params[system] = system

    messages = client.messages.create(**params)
    return messages.content[0].text
```

```python
# Program logic

messages =[]

add_user_messages(messages, "Generate a one sentence movie idea")

answer = chat(messages,temperature=0.70)

answer
```
