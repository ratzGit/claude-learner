```python
#Load headers

from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
# Helper functions

def add_user_messages(messages, text):
    user_messages = {"role":"user", "content":text}
    messages.append(user_messages)

def add_assistant_messages(messages, text):
    assistant_messages = {"role":"assistant", "content":text}
    messages.append(assistant_messages)

def chat(messages, seq=None):

   params = {
        "messages": messages,
        "max_tokens": 1000,
        "model": model,
   }

   if seq:
       params["stop_sequences"]= seq

   answer = client.messages.create(**params)
   return answer.content[0].text


    
```

```python
# Program Logic

messages = []
seq = ["5","3,4"]

add_user_messages(messages,"Count from 1 to 10")

chat(messages, seq)
```
