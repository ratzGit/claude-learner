```python
# Load headers

from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Helper functions

def add_user_messages(messages,txt):
    user_messages = {"role":"user", "content":txt}
    messages.append(user_messages)

""" def add_system_messages(messages,txt):
    system_messages = {"role":"system", "content":txt}
    messages.append(system_messages) """

def chat(messages,sysPrompt):
    response = client.messages.create(
        max_tokens=1000,
        messages=messages,
        model=model,
        system=sysPrompt
    )
    
    return response.content[0].text
```

```python
message = []

sysPrompt = """
The user needs only the commands without any JSON formatting. 
This will help the user copy the commands quickly. 
Only send the commands as output.
"""

prompt = """
Generate three different sample AWS CLIcommands. Each should be very short and don't use newline characters in between the commands.
"""

add_user_messages(message, prompt)


chatText = chat(message, sysPrompt)
chatText.strip()
```
